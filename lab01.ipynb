## Task 7 - Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

- Khi sử dụng dim=0, các phép tính được thực hiện theo từng cột. Với tensor có shape (3, 4), mean và std tính trên 3 phần tử của mỗi cột nên kết quả có 4 phần tử (shape (4,)). cumsum cộng dồn từ trên xuống dưới trong từng cột nên kết quả vẫn có shape (3, 4).

- Khi thay bằng dim=1, các phép tính được thực hiện theo từng hàng. mean và std tính trên 4 phần tử của mỗi hàng nên kết quả có 3 phần tử (shape (3,)). cumsum cộng dồn từ trái sang phải trong từng hàng nên kết quả vẫn có shape (3, 4).

## Task 9 - Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    
print("Learned coefficients:")
print(w)
```

- Hệ số 1: 1.0138 gần 1 → học khá chính xác. 

- Hệ số 2: 2.0306 gần 2 → học khá chính xác.

- Hệ số 3: 2.3522 còn lệch khá nhiều so với 3.

- optimizer.zero_grad(): xóa gradient của vòng lặp trước, vì Pytorch cộng dồn gradient, nên mỗi epoch cần xóa gradient cũ trước khi tính gradient mới

- mse.backward(): Tính gradient của mse đối với w

- optimizer.step(): cập nhật giá trị w dựa trên gradient vừa tính


## Task 10 - Giải hệ phương trình trình bằng Autorgrad

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

### a. Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.

```python
import torch

a = torch.rand(24)

print("\nOriginal tensor:")
print(a)

print("\nReshape to 4 x 6:")
print(a.reshape(4, 6))

print("\nReshape to 2 x 12:")
print(a.reshape(2, 12))

print("\nReshape to 2 x 3 x 4:")
print(a.reshape(2, 3, 4))
```

### b. Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.

```python
import torch

A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("A+B")
print(A + B)

print("A*B")
print(A * B)

print("A/B")
print(A / B)

print("\nMean A:")
print(torch.mean(A, dim=0))

print("\nStandard deviation A:")
print(torch.std(A, dim=0))

print("\nMean B:")
print(torch.mean(B, dim=0))

print("\nStandard deviation B:")
print(torch.std(B, dim=0))
```

### c. Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = 2 * x ** 2 + 5 * x + 3

y.backward()

print("x = ", x)
print("y = ", y)
print("x.grad   = ", x.grad)
```

### d. Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch
polynomial = np.poly1d([2, -3, 5])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    
print("Learned coefficients:")
print(w)
```
